```python
import sqlite3
import sys

try:
  with sqlite3.connect("ConFechas_timestamp.db") as conexion:
    cursor = conexion.cursor()
    inst = "CREATE TABLE amigo (clave INETEGER PRIMARY KEY, nombre TEXT NOT NULL, fecha_registro TIMESTAMP);"
    cursor.execute(inst)
    print("Tabla creada papu")
except sqlite3.Error as e:
  print(e)
except:
  print(f"Se produjo el siguiente error: {sys.exc.info()[0]}")
finally:
  if(conexion):
    conexion.close()
    print("Se cerro la conexion")
```

```python
from sqlite3 import Error
import datetime

try:
  with sqlite3.connect("ConFechas_timestamp.db") as conexion:
    cursor = conexion.cursor()
    datenow = datetime.datetime.combine(datetime.date.today(), datetime.datetime.min.time())
    vals = {"clave": 30, "nombre":"Patrick", "fecha_registro": datenow}
    cursor.execute("INSERT INTO amigo VALUES (:clave, :nombre, :fecha_registro)", vals)
    print("Registro insertado papu")
except Error as e:
  print(e)
except:
  print(f"Se produjo el siguiente error: {sys.exc.info()[0]}")
finally:
  if(conexion):
    conexion.close()
    print("Se cerro la conexion")
```

```python
try:
  with sqlite3.connect("ConFechas_timestamp.db", detect_types = sqlite3.PARSE_DECLTYPES | sqlite3.PARSE_COLNAMES) as conexion:
    cursor = conexion.cursor("ConFechas_timesatmp.db")
    cursor.execute("SELECT clave, nombre, fecha_registro FROM amigo")
    registros = cursor.fetchall()

    if registros:

      for clave, nombre, fecha_registro in registros:
        print(f"Clave: {clave}, tipo de dato: {type(clave)}")
        print(f"Nombre: {nombre}, tipo de dato {type(nombre)}")
        print(f"Fecha de registro: {fecha_registro}, tipo de dato: {type(fecha_registro)}")
    else:
      print("No se encontraron registros")
except Error as e:
  print(e)
except Exception:
  print(f"Se produjo el siguiente error: {sys.exc_info()[0]}")
finally:
  if(conexion):
    conexion.close()
    print("Se cerro la conexion")
```
