```python
import openpyxl

book = openpyxl.Workbook()
sheet = book.active
sheet.title = "Productos"
print("Archivo de Excel creado")
```

```python
sheet["A1"] = "Producto"
sheet["B1"] = "Precio"
sheet["C1"] = "Cant."

sheet["A2"] = "Cuaderno"
sheet["B2"] = "$45"
sheet["C2"] = "10"

sheet["A3"] = "Pluma"
sheet["B3"] = "$12"
sheet["C3"] = "25"

sheet["A4"] = "Lapiz"
sheet["B4"] = "$8"
sheet["C4"] = "30"

sheet["A5"] = "Borrador"
sheet["B5"] = "$10"
sheet["C5"] = "15"

book.save("prods.xlsx")
print("Archivo guardado")

book.close()
print("Archivo cerrado")
```

```python
book = openpyxl.load_workbook("prods.xlsx")
sheet = book["Productos"]

for fila in sheet.iter_rows(values_only=True):
    print(fila)
```

```python
prod = input("Nombre del prod: ")

for fila in sheet.iter_rows(values_only=True):
  if fila[0] == prod:
    print(f"Precio: {fila[1]}")
    print(f"Cant.: {fila[2]}")
```
