```python
class stack:
  def __init__(self):
    self.items = []
  def is_empty(self):
    return self.items == []
  def push(self, item):
    self.items.insert(0, item)
  def pop(self):
    return self.items.pop(0)
  def print_stack(self):
    print(self.items)

pila = stack()

pila.push('a')
pila.push('b')
pila.push('c')
pila.push('d')
pila.push('e')
pila.print_stack()
pila.pop()
pila.print_stack()
pila.pop()
pila.print_stack()
```

```python
class pila:
  def __init__(self):
    self.items = []

  def esta_vacia(self):
    return self.items == []

  def incluir(self, item):
    self.items.append(item)

  def extraer(self):
    return self.items.pop()

s = pila()

s.incluir("Hola")
s.incluir("Angel")
s.incluir("es")
s.incluir(67)
s.incluir("gay")
print(s.extraer())
```

```python
stack = []
SEPARA = ("-" * 20) + "\n"

def main():
  print('''1 Apilar elemento (entero)")
2 Desapilar elemento
3 Mostrar pila
4 Salir
''')
  print(SEPARA * 1)
  option = input("Elija una opcion: ")

  if str(option) == "1":
    elemento = input("Ingrese el elemento a apilar: ")
    stack.append(elemento)
    print("**- Elemento apilado -** \n")
    main()

  elif str(option) == "2":
    if len(stack) > 0:
      print(f"**- Elemento {stack.pop()} desapilado -**")
      main()
    else:
      print("**- La pila esta vacia -**")
      main()

  elif str(option) == "3":
    for i in reversed(range(len(stack))):
      print("Pila: ", stack[1])
      main()

  elif str(option) == "4":
    exit()

  else:
    print("Opción no válida")
    main()

main()
```

```python
class Cola:

  def __init__(self):
    self.items = []

  def estaVacia(self):
    return self.item == []

  def agregar(self, item):
    self.items.insert(0, item)

  def avanzar(self):
    return self.items.pop()

  def tamano(self):
    return len(self.items)

q = Cola()
q.agregar(4)
q.agregar("Angel")
q.agregar(True)
q.tamano()
```

```python
cola = []
separa = ("-" * 20) + "\n"

def main():
  print("1 Insertar cola")
  print("2 Borrar en cola")
  print("3 Listar cola")
  print("4 Salir")
  print(separa * 1)

  option = input("Elija una opcion: ")

  if str(option) == "1":
    elemento = input("Ingrese el elemento a apilar: ")
    cola.append(elemento)
    print("**- Elemento apilado -** \n")
    main()

  elif str(option) == "2":
    if len(cola) > 0:
      print(f"**- Elemento {cola.pop()} desapilado -** \n")
      main()
    else:
      print("**- La cola esta vacia -** \n")
      main()

  elif str(option) == "3":
    count = 0
    for i in cola:
      print(f"Cola {count + 1}: ", i)
      count += 1
    main()

  elif str(option) == "4":
    print("**- Programa terminado -**")
    exit()

  else:
    print("Opción no válida")
    main()

main()
```
