```python
cola = list()

for cant in range(5):
  new = input("Nombre a agrerar: ")
  cola.append(new)

print(f"\nSe han agregado {len(cola)} elementos a la cola\n")
for elemento in cola:
  print(elemento)

print(f"\nActualmente quedan {len(cola)} elementos en la cola\n")

print(f"Procedemos a retirar la cola")

while cola:
  print(cola.pop())

print(f"\nActualmente quedan {len(cola)} elementos en la cola\n")
```

```python
import collections

cola = collections.deque()

for cant in range(5):
  new = input("\nNombre del agregado: ")
  cola.append(new)

print(f"\nSe han agregado {len(cola)} elementos a la cola\n")

for elemento in cola:
  print(elemento)

print(f"\nActualmente quedan {len(cola)} elementos en la cola\n")

print("\nProcedemos a retirarlos de la cola")

while cola:
  print(cola.popleft())

print(f"\nActualmente quedan {len(cola)} elementos en la cola\n")
```
