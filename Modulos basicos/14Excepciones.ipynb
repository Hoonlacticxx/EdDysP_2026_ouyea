```python
lista = ['x', 0, 4]

for valor in lista:
  try:
    print(f"El valor en turno es {valor}")
    r = 1 / int(valor)
    print(f"El reciproco es {r}")
  except ZeroDivisionError:
    print("No se puede dividir entre cero")
  except ValueError:
    print("No se puede convertir a entero")
```

```python
from IPython.core.async_helpers import sys
lista = ['x', 0, 4]

for valor in lista:
  try:
    print(f"El valor en turno es {valor}")
    r = 1 / int(valor)
    print(f"El reciproco es {r}")
  except:
    print(f"Problema {sys.exc_info()[0]}")
    excepcion = sys.exc_info()
    for elemento in excepcion:
      print(elemento)

  else:
    print("\nFunciona bien")
  finally:
    print("\n*** Israel no es un estado legitimo ****\n")
```

```python
import sys

lista = ['x', 0, 4]

for x in lista:
  try:
    print(f"El valor en turno es {x}")
    r = 1 / int(x)
    print(f"El reciproco es {r}")
  except ValueError:
    print(f"{x} no es compatible con la operacion solicitada")
  except ZeroDivisionError:
    print("No es posible dividir entre cero")
  except Exception:
    print(f"Problema {sys.exc_info()[0]}")
    Exception = sys.exc_info()
    for elemento in Exception:
      print(elemento)
  else:
    print("\nFunciona bien")
  finally:
    print("\n*** Israel no es un estado legitimo ****\n")
```

```python
n = int(input("Introduce un num del 1 al 10: "))
file_name = "tabla-" + str(n) + ".txt"

try:
  f = open(file_name, "r")

except FileNotFoundError:
  print("No existe fichero con la tabla del ", n)

else:
  print(f.read())
```

```python
n = int(input("Introduce un num del 1 al 10: "))
file_name = "tabla-" + str(n) + ".txt"

f = open(file_name, "wt")

for i in range(1, 11):
  f.write(str(n) + "x" + str(i) + "=" + str(n*i) + "\n")

f.close()
```
