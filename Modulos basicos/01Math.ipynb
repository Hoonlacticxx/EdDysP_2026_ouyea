```python
import math

vF = float(input("Introduce un número con decimal: "))

print(f"El entero anterior de {vF} es {math.floor(vF)}")

print(f"El sig. entero de {vF} es {math.ceil(vF)}")

print(f"La parte entera de {vF} es {math.trunc(vF)}")
```

```python
vN = int(input("Dame un num. "))

print(f"La raíz cuadrada de {vN} es {math.sqrt(vN)}")

print(f"El factorial de {vN} es {math.factorial(vN)}")
```

```python
p = int(input("Valor entero: "))

print(f"{p} elevado al cuadrado es {math.pow(p, 2)}")

print(f"El valor de Pi es {math.pi}")
```
