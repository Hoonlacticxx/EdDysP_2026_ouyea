```python
data = {'nombre':['Maria', 'Luis', 'Carmen'],
        'edad':[18, 22, 20],
        'Matematicas':[8.5, 7, 3.5],
        'Economia':[8, 6.5, 5],
        'Programacion':[6.5, 4, 9]}

df = pd.DataFrame(data)
df1 = df.melt(id_vars=['nombre', 'edad'],
              var_name='asignatura', value_name='nota')
print(df1)
```

```python
print(df1.pivot(index='nombre', columns='asignatura', values='nota'))
```

```python
notas = pd.Series([87, 100, 85, 60, 78])
print(type(notas))
print(notas, '\n')

eq = pd.Series(100, range(6))
print(type(eq))
print(eq)
```

```python
print('notas: ')
print(f'{notas}')

print(f'Cant = {notas.count()}')
print(f'Media = {notas.mean()}')
print(f'Mediana = {notas.median()}')
print(f'Min = {notas.min()}')
print(f'Max = {notas.max()}')
print(f'Desv. Estándar = {notas.std()}')

print('Sumarizacion descr.')
print(notas.describe())
print(type(notas.describe()))
```
