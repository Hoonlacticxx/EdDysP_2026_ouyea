```python
data = np.random.randint(1, 200, size=(1, 8))
print("Los datos son:")
print(data)

min = data.min()
max = data.max()

print(f"\nLos datos que tiene un rango de {max - min}, abarcando desde el {min} hasta el {max}")
```

```python
min = np.amin(data)
max = np.amax(data)

print(f"\nLos datos que tiene un rango de {max - min}, abarcando desde el {min} hasta el {max}")

primer_cuartil = np.percentile(data, 25)
tercer_cuartil = np.percentile(data, 75)
mediana = np.median(data)

print(f"\nEl primer cuartil es: {primer_cuartil}")
print(f"El tercer cuartil es: {tercer_cuartil}")
print(f"\nLa mediana es: {mediana}")
```

```python
meadia = data.mean()
desv_est = data.std()
var = data.var()

print(f"\nLa media es: {meadia}")
print(f"La desviación estándar es: {desv_est}")
print(f"La varianza es: {var}")
#
```

```python
cals = np.array([92, 100, 72, 88, 95])
prom = np.mean(cals)
max = np.max(cals)
min = np.min(cals)
cant = np.sum(cals >= 70)

print("Promedio: ", prom)
print("Máximo: ", max)
print("Mínimo: ", min)
print("Cantidad de aprobados: ", cant)

if prom >= 70:
    print("El alumno está aprobado")
else:
    print("El alumno está desaprobado")
```

```python
temps = np.array([18, 21, 25, 19, 23, 27, 30])
prom = np.mean(temps)
dias_encima = np.sum(temps > prom)
farenheit = temps * 9/5 + 32

print("Promedio: ", prom)
print("Días con temperatura por encima del promedio: ", dias_encima)
print("Temperaturas en grados Fahrenheit: ", farenheit)
```

```python
ventas = np.array([129, 258, 388, 516, 18])
ingresos = np.sum(ventas)
prom = np.mean(ventas)
max = np.max(ventas)
min = np.min(ventas)

print("Ingresos totales: ", ingresos)
print("Promedio de ventas: ", prom)
print("Venta máxima: ", max)
print("Venta mínima: ", min)
```
