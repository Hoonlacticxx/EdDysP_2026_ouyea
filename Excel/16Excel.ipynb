```python
import datetime
import openpyxl
import random

book = openpyxl.Workbook()

hoja = book["Sheet"]
hoja.title = "Primera"

hoja["B1"].value = "Listado de nums"

for r in range(2, 12):
  hoja.cell(row=r, column=2).value = random.randrange(100)

hoja["C12"].value = "=SUM(B2:B11)"

hoja["C15"].value = "Hoy es: "
hoja["D15"].value = datetime.datetime.now().strftime("%d/%m/%Y")

book.save("listado.xlsx")

print("Libro creado, papupro")
```
