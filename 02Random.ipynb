```python
import random

print(f"Numero aleatorio de entre el 0 al 50. {random.randrange(51)}")

print(f"Numero par aleatorio de entre el 0 al 50. {random.randrange(2, 51, 2)}")

print(f"Numero aleatorio de entre el 1 al 20. {random.randint(1, 20)}")

print(f"Numero aleatorio entre 0.0 y 1.0. {random.random()}")
```

```python
lPrueba = [10, 20, 30, 40, 15, 25, 35, 45]
print(f"La lista de prueba es: {lPrueba}")

print(f"Numero aleatorio de la lista de prueba. {random.choice(lPrueba)}")

random.shuffle(lPrueba)
print(f"La lista de prueba mezclada es: {lPrueba}")
```
